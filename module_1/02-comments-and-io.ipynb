# comments
**comments** -->> generally/often used to refer to a part/section of code to let other developers know what that code actually do. may also be used by the same developer to revise the code's functionality.

example >>  
```python
    # return your age assuming we're in the year 2026
    dob = int(input('enter your year of birth:\t))
    print(f"hey bud! assuming that we're in the year 2026\nyou're about {2026-dob} years old.")
```
actually executed the code below:

````python
# return your age assuming we're in the year 2026
dob = int(input('enter your year of birth:\t'))
print(f"hey bud! assuming that we're in the year 2026\nyou're about {2026-dob} years old.")
````

### 1. single line comment
in python we use **#** for single line commnets
example >>  
```python
    # comment line
```
**2. multi line comment**
python doesn't have the concept of multi-line comment. so instead we use **DOC STRING**  
### **doc string**
example >>  
```python
    '''
    doc string
    this could also be used as multi-line comments in python
    we generally use this to explain how the code is to be used
    '''
```
