# variables (or python variables) -> container to store data
variable types: (normally of these two types throughout most of the programming languages):  
**1. statistically typed**  
**2. dynamically typed**

for example  
> *C language is statistically typed*
```c
    int x = 5
```

> *while python is dynamically typed*
```python
    x = 5
```

````python
# creating a variable in python
name = "@jkmloom"
code_year = 2026
weight_kg = 88.75
````

python considers evey variable as an object. this object has got an address and we can find this adddress using an in-built function/method -> `id`

````python
# creating an variable
age = 23
# finding the object's address
print(id(age))
````

> the object contains the following info:  

**considering object id: 110994320776424**  
type            -> integer  
value           -> 23  
reference count -> 1

````python
# if we've got 2 variables pointing to the same value then the variables will be sharing the same object id
x = 5
print(id(x))
y = 5
print(id(y))
````

> if multiple variables are pointing to the same value then the reference count changes accordengly:  

for example  **considering object id: 110994320775848** >>  
type -> integer  
value -> 5  
reference count -> 2

> **rules for naming a variable** *(in python)*

1. allowed characters -->>  
    i.  a/A - z/Z (alphabets)  
    ii. _ (underscore/s)  
    iii. 0, 1, 2, 3, .. (numerical values)  
[!, @, #, $, %, ^, etc. are not allowed]  

2. first character of the variable must be a alphabet of an underscore  
[we can't use numerical values like 0, 1, 2, 3, ..]  

3. variable names are case sensative (a and A are different)  

4. avoid using keywords as variable names (not allowed)  
**keywords** -->> reserved words whose meanings are already assigned to the computer.  

5. variable name should be a single word.

````python
# for example
variable_1 = "this is an example of a valid variable name in python"
_2variable = "this is another example of a valid variable name in python"

a = 23
A = 24
print(f"a = {a} and A = {A}")
````
