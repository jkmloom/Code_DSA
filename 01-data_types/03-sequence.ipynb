# **03. sequence**
- collection of elements >> python allows different data types
- list is ordered >> it has a concept of index
- it's mutable
- *python doesn't have array in-built but we make use of list*

```python
students = ['mohan', 'vishwa', 'shivani', 'sakshi']
print(students)
print(type(students))
```

```python
list_2 = list(students) # list() >> constructor method
print(list_2)
```

```python
list_3 = list('vishwa')
print(list_3)
```

```python
list_4 = list((3, 4, 'jkmloom'))
print(list_4)
```

```python
print(type(list_4))
```
