# **02. numerical data types**

```python
x=5
# x is a variable storing an integer
```

```python
price = 26.25
# price is a variavle storing an floating point value
```

```python
print(f"integer: {x}")
print(f"floating point: {price}")
```

```python
# for a large number, say: 3 * 10^6 = 3,000,000
num = 3e6
# here 'e' = 10^e
print(f"{num}")
```

```python
# complex number
# for 3 + 4i we present it in python as 3 + 4j
z = 4 + 5j
print(f"complex number: {z}")
```

```python
# check data type of a given value
print(type(z))
```

### **in-built functions**

```python
# abs() >> returns positive values
print(abs(-5))
```

```python
# pow(x, y) >> for x^y
print(pow(2, 3))
```

```python
# round() >> returns the nearest integer
print(round(4.75))
print(round(4.32))
```

```python
# min() >> finds minimum value
# max() >> finds maximum value
nums = [3, 1, 5, 7, 9, 1]
print(min(nums))
print(max(nums))
```

```python
# sum()
print(sum(nums))
```
